```python
#8 Find the repetitive elements in the list [1,2,3,4,56,1,22,23,33,23, 56].
numbers=[1,2,3,4,56,1,22,23,33,23, 56]
repetitive = []
for num in numbers:
    if numbers.count(num) > 1 and num not in repetitive:
        repetitive.append(num)

print("Repetitive elements:", repetitive)
```

```python
#28 . Create a new list containing only prime numbers from a given list.
lst = [2, 3, 4, 5, 6, 7, 11, 15]
def is_prime(n):
    if n < 2: return False
    for i in range(2, int(n**0.5)+1):
        if n % i == 0: return False
    return True
primes = [x for x in lst if is_prime(x)]
print("Primes:", primes)
```

```python
#36.Rearrange a list so that positive and negative numbers alternate.
lst = [1, 2, -3, -4, 5, -6]
pos = [x for x in lst if x > 0]
neg = [x for x in lst if x < 0]
result = []
for p,n in zip(pos,neg):
    result.extend([p,n])
print("Alternating:", result)
```

```python
#42.Find All Anagram Groups in a List of Words.
lst = ["eat","tea","tan","ate","nat","bat"]
anagrams = {}
for word in lst:
    key = "".join(sorted(word))
    anagrams.setdefault(key, []).append(word)
print("Anagram groups:", list(anagrams.values()))
```
